#### Constructing Columns on Imputed New and OLD

```python
new_imputed_elems = [element for element in df.columns if 'Imputed New' in element]
old_imputed_elems = [element for element in df.columns if 'Imputed Old' in element]
```

#### Let's try a common geochemical data transformation

```python
# adding them to the dataframe
df[new_imputed_elems] = df[new_imputed_elems]*10000 #convert from wt% to ppm
df[old_imputed_elems] = df[old_imputed_elems]*10000 #convert from wt% to ppm
```

```python
# transforming them
from scipy.stats import gmean

# new elements
clr_names = []

for name in new_imputed_elems:

    name_string = name+'_CLR'

    df[name_string] = np.log(df[name]/gmean(df[new_imputed_elems], axis = 1))

    clr_names.append(name_string)

# old elements
clr_names = []

for name in old_imputed_elems:

    name_string = name+'_CLR'

    df[name_string] = np.log(df[name]/gmean(df[old_imputed_elems], axis = 1))

    clr_names.append(name_string)
```

### Reading Transformed data

```python
# Loading Data
url = "https://raw.githubusercontent.com/kwanda2426/projects/main/geospatial/Hw_2/CLR_Imputed_merged_chem_KM.csv"
df = pd.read_csv(url)
df.head()
```

```python
# formatting column names
df.columns = df.columns.str.replace(' ', '_', regex=False)
df.columns = df.columns.str.replace('New', 'NEW', regex=False)
df.columns = df.columns.str.replace('Old', 'OLD', regex=False)
df.columns
```

```python
#old cols
old_cols_temp = [i for i in df.columns if 'Imputed_OLD' in i and 'CLR' in i] # with full name
old_cols = [i[8:] for i in df.columns if 'Imputed_OLD' in i and 'CLR' in i] # without Imputed

#New cols
new_cols_temp = [i for i in df.columns if 'Imputed_NEW' in i and 'CLR' in i] # with full name
new_cols = [i[8:] for i in df.columns if 'Imputed_NEW' in i and 'CLR' in i] # without Imputed
```

```python
# final cols and df with Imputed and CLR transformed columns
final_cols = old_cols_temp + new_cols_temp
df_1 = df[final_cols]
df_1.columns = df_1.columns.str.replace('Imputed_', '', regex=False)
```

```python
df_1.columns
```

```python
# elements in both old and new after CLR
pairs = []
new_elements = [j.split('_')[1] for j in new_cols]
elements_to_be_used = []
print('These are the overlapping elements')
for i in old_cols:
  old_element = i.split('_')[1]
  if old_element in new_elements:
    pair = ['OLD_'+old_element+'_CLR','NEW_'+old_element+'_CLR']
    elements_to_be_used.append('OLD_'+old_element+'_CLR')
    elements_to_be_used.append('NEW_'+old_element+'_CLR')
    print(old_element)
    pairs.append(pair)
print('There are {} pairs of elements'.format(len(pairs)))
```

```python
# Double-check each name appears in df.columns: if not, an error will be raised
for old_col, new_col in pairs:
    assert old_col in df_1.columns and new_col in df_1.columns, f"Missing: {old_col} or {new_col}"
```

```python
# cols to plot and final df
all_cols = elements_to_be_used
df_final = df_1[all_cols]
df_final.head()
```

# **Visualisation**

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Step 1: Identify common elements
old_cols = [col for col in df_final.columns if col.startswith('OLD_')]
new_cols = [col for col in df_final.columns if col.startswith('NEW_')]

# Extract element-condition keys
elements = [col.replace('OLD_', '') for col in old_cols if f'NEW_{col.replace("OLD_", "")}' in new_cols]

# Step 2: Loop through each element and plot
for elem in elements:
    x = df_final[f'OLD_{elem}']
    y = df_final[f'NEW_{elem}']

    plt.figure(figsize=(6, 6))
    plt.scatter(x, y, alpha=0.7, label='Data Points')

    # Regression line without label
    m, b = np.polyfit(x, y, 1)
    plt.plot(x, m*x + b, color='red')  # No label here

    # Axis limits and labels
    min_val = min(x.min(), y.min())
    max_val = max(x.max(), y.max())
    plt.xlim(min_val, max_val)
    plt.ylim(min_val, max_val)
    plt.xlabel(f'OLD {elem}')
    plt.ylabel(f'NEW {elem}')
    plt.title(f'Scatterplot: OLD vs NEW {elem}')
    plt.grid(True)
    plt.tight_layout()
    plt.show()
```
