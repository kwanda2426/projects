```python
from rdkit import Chem
from rdkit.Chem import Draw
from rdkit.Chem import Descriptors

import numpy as np
import pandas as pd
```

# **Import Data**

```python
# Reading data
url = "https://raw.githubusercontent.com/kwanda2426/projects/main/ML_In_Chem/solubility.csv"
df = pd.read_csv(url)
df.head()
```

# **Molecular Information**

# **Drawing molecules**

```python
df.columns
```

```python
# first six molecules
df_6 = df[['Compound ID','smiles']].iloc[:6]
df_6
```

```python
compound_ = df_6['Compound ID'].tolist()
smiles_ = df_6['smiles'].tolist()

for i in range(len(compound_)):
  print(compound_[i])
  print('')
  print(smiles_[i])
smiles = smiles_[i] # Replace with your SMILES notation
mol = Chem.MolFromSmiles(smiles)
Draw.MolToImage(mol)
```

```python
#molecular weight
atomic_weight = Descriptors.MolWt(mol)
print(f" Atomic Weight : { atomic_weight } g/ mol ")
```

```python
# Heavy Atom Count
heavy_atom_count = Descriptors.HeavyAtomCount(mol)
print(f" Heavy Atom Count : { heavy_atom_count }")
```

```python
# Number of Valence Electrons:
valence_electrons = Chem.Descriptors.NumValenceElectrons(mol)
print(f" Number of Valence Electrons : { valence_electrons }")
```
